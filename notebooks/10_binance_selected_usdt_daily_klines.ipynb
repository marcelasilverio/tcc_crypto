# Binance Spot daily klines for the selected USDT universe

This notebook collects closed `1d` Binance Spot klines for the approved 40-pair universe. The requested period is fixed from `2017-01-01 00:00:00 UTC` through `2026-09-01 23:59:59.999 UTC`. Each pair starts at its first available Binance candle; observations are not synthesized before listing or across missing intervals.

For each pair, the notebook creates immutable canonical `.csv.gz` data, a derived Excel review workbook, and a JSON metadata sidecar under `data/raw/binance/spot/1d/`. A timestamped manifest summarizes the execution. The existing hourly collection remains unchanged.

## Research limitation

The universe was selected ex post using activity observed through 2026-09-01 and current symbol status. The fixed universe must not be presented as historically known at earlier portfolio decision timestamps.

```python
import importlib
from pathlib import Path
import sys

import pandas as pd


def find_repository_root(start: Path | None = None) -> Path:
    """Find the repository root from a notebook or project directory."""
    current = (start or Path.cwd()).resolve()
    for candidate in (current, *current.parents):
        if (candidate / "AGENTS.md").exists() and (candidate / "pyproject.toml").exists():
            return candidate
    raise FileNotFoundError("Could not locate the repository root.")


REPOSITORY_ROOT = find_repository_root()
SOURCE_DIRECTORY = REPOSITORY_ROOT / "src"
if str(SOURCE_DIRECTORY) not in sys.path:
    sys.path.insert(0, str(SOURCE_DIRECTORY))

import tcc_crypto.binance_kline_collection as collection_module

collection_module = importlib.reload(collection_module)
load_collection_config = collection_module.load_collection_config
run_collection = collection_module.run_collection

CONFIG_PATH = REPOSITORY_ROOT / "configs/binance_selected_usdt_daily_collection.json"
config = load_collection_config(CONFIG_PATH)
print(f"Repository root: {REPOSITORY_ROOT}")
print(f"Configuration: {CONFIG_PATH}")
```

```python
configuration_summary = pd.DataFrame(
    [
        ("Collection ID", config.collection_id),
        ("Market", config.market),
        ("Interval", config.interval),
        ("Requested start (UTC)", config.start_time_utc.isoformat()),
        ("Requested end (UTC)", config.end_time_utc.isoformat()),
        ("Selected pairs", len(config.symbols)),
        ("Output directory", str(REPOSITORY_ROOT / config.output_directory)),
        ("Source endpoint", f"{config.base_url}{config.kline_path}"),
    ],
    columns=["Field", "Value"],
)
display(configuration_summary)
display(
    pd.DataFrame(
        {
            "Selection Rank": range(1, len(config.symbols) + 1),
            "Trading Pair": config.symbols,
        }
    )
)
```

## Run the collection

The following cell collects all 40 approved pairs. Collection resumes after the last page persisted in each pair checkpoint. Failed pairs are recorded in the manifest without deleting successful outputs.

```python
collection_results, manifest_path = run_collection(
    config,
    REPOSITORY_ROOT,
    progress_callback=print,
)
display(collection_results)
print(f"Collection manifest: {manifest_path}")
```

```python
status_summary = (
    collection_results["Status"]
    .value_counts(dropna=False)
    .rename_axis("Status")
    .reset_index(name="Pairs")
)
display(status_summary)

failed = collection_results.loc[collection_results["Status"] == "FAILED"]
pairs_requiring_gap_review = collection_results.loc[
    collection_results["Missing Days"].fillna(0) > 0
]
if not pairs_requiring_gap_review.empty:
    print("Pairs with missing daily intervals requiring review:")
    display(pairs_requiring_gap_review[["Trading Pair", "Missing Days", "Workbook"]])

pairs_with_shortened_candles = collection_results.loc[
    collection_results["Shortened Candles"].fillna(0) > 0
]
if not pairs_with_shortened_candles.empty:
    print("Pairs with shortened historical candles requiring review:")
    display(
        pairs_with_shortened_candles[
            ["Trading Pair", "Shortened Candles", "Workbook"]
        ]
    )

if not failed.empty:
    display(failed[["Trading Pair", "Error"]])
    raise RuntimeError(
        f"Collection finished with {len(failed)} failed pair(s). "
        "Review the manifest and rerun the collection cell."
    )

print("All requested pairs were collected or already existed.")
```

## Interpretation and next checks

A missing UTC day inside the observed span is recorded as `REVIEW`; it is not filled in the canonical raw CSV. A candle whose close remains inside its nominal 24-hour window but occurs before the final millisecond is preserved and marked `REVIEW`, with details in `Nonstandard Durations`. A close before the open or beyond the nominal daily window remains a fatal validation failure. Time before a pair's first observed candle is treated as pre-listing history rather than missing data.
