```python
from IPython.display import display
from tcc_crypto.workflows import exploration as workflow

RUN_ID = None  # Set an existing run identifier to inspect a saved execution.
OVERWRITE = False
context = workflow.open_exploration('future-high', run_id=RUN_ID)
display(workflow.paths_table(context))
```

## 3. Calculate and inspect results


```python
result = workflow.calculate(context, inputs)
display(workflow.table_preview(result, 'outcomes'))
```

## 6. Save the calculated tables
